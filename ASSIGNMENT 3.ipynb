```python
def sides():
    s1=input("Enter your number")
    s2=input("Enter your number")
    s3=input("Enter your number")
    
    if(s1==s2==s3):
        print("Triangle is equilateral")
    
    elif (s1==s2 or s2==s3 or s3==s1):
        print("Triangle is isoceles")
    else:
        print("Triangle is scalene")

sides()
```
